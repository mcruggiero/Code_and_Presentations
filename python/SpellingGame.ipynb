```python
import random
from gtts import gTTS
from IPython.display import Audio, display

# 🎧 Text-to-speech function using gTTS
def say(word):
    tts = gTTS(text=word, lang='en')
    tts.save("temp.mp3")
    return Audio("temp.mp3", autoplay=True)

# ✏️ Expanded 2nd Grade Words
words_2nd = [
    "about", "again", "animal", "answer", "basket", "because", "began", "Belmont", "book", "boxes", "bright", "buy",
    "camp", "catch", "center", "chick", "city", "class", "clean", "dinner", "done", "drink", "each", "eight", "every",
    "father", "feeling", "fight", "first", "flower", "foot", "found", "fuzzy", "gone", "grabbed", "grass", "horse",
    "house", "jelly", "know", "last", "laugh", "lunch", "many", "mother", "nest", "once", "pass", "please", "queen",
    "rock", "school", "star", "spell", "swimming", "think", "this", "trapped", "until", "wall", "were", "yellow",
    "window", "table", "friend", "happy", "pencil", "paper", "baby", "little", "apple", "water", "name", "good", "play",
    "rain", "sun", "tree", "fast", "slow", "day", "night", "sleep", "wake", "cold", "warm", "family"
]

# 🧠 5th Grade Words
words_5th = [
    "absence", "accommodation", "accidentally", "agriculture", "angel", "appearance", "approximately", "association",
    "bankruptcy", "beneficial", "breathe", "calendar", "category", "chemistry", "civilization", "comparative",
    "competition", "confirmation", "conscientious", "consequently", "controversial", "curriculum", "delegate",
    "develop", "difference", "disastrous", "distinguished", "dormitory", "education", "especially", "essential",
    "etiquette", "exaggeration", "executive", "experience", "extraordinary", "foundation", "fragile", "guidance",
    "gymnasium", "heroic", "hygiene", "illustrate", "immense", "inability", "indispensable", "innocence", "intellect",
    "interpretation", "irresistible", "jealousy", "laborious", "lightning", "magnificence", "maturity", "miniature",
    "mischievous", "misspelled", "mortgage", "neighbor", "nuisance", "obstacle", "occasionally", "offensive",
    "opinion", "organization", "pageant", "pamphlet", "parachute", "peaceable", "permanent", "pigeon", "poison",
    "politician", "possibility", "prejudice", "preparation", "priority", "procedure", "procession", "proposition",
    "psychology", "publicity", "qualities", "questionnaire", "reference", "register", "religious", "remembrance",
    "requirement", "restaurant", "ridiculous", "satisfactorily", "senator", "sensibility", "significance", "solution",
    "sophomore", "specific", "specimen", "strenuous", "substantial", "successful", "sufficient", "summarize",
    "superintendent", "syllable", "symphony", "technique", "temperature", "tournament", "tyranny", "unanimous",
    "vacillate", "vengeance", "villain", "vivacious", "voracious"
]

# 🧸 Easy Recovery Words (Grades 1–3)
recovery_words = [
    "cat", "dog", "run", "jump", "book", "red", "blue", "mom", "dad", "yes", "no", "car", "ball", "sun", "bed",
    "hat", "pen", "box", "toy", "fish", "cake", "milk", "shoe", "bird", "tree", "egg", "cup", "top", "bag", "bell"
]

# 🔘 User Menu
grade = input("Select spelling level (2 or 5): ").strip()

if grade == '2':
    words = words_2nd
else:
    words = words_5th  # include 2nd grade words as review for 5th graders

try:
    num_questions = int(input("How many questions would you like to play? (e.g. 5, 10, 15): ").strip())
except ValueError:
    print("Invalid number, defaulting to 10 questions.")
    num_questions = 10

# 🎮 Game Start
print("\n🎧 Starting Speak and Spell! 🎧\n")
random.shuffle(words)
score = 0

for word in words[:num_questions]:
    print("Listen carefully...")
    display(say(word))
    guess = input("Type what you heard ➜ ").strip().lower()
    
    if guess == word:
        print("✅ Correct!\n")
        score += 1
    else:
        print(f"❌ The correct word was '{word}'. Let's try an easier one!\n")
        
        # 🎁 Bonus recovery round
        bonus_word = random.choice(recovery_words)
        print("Bonus word! Listen again...")
        display(say(bonus_word))
        bonus_guess = input("Type what you heard ➜ ").strip().lower()
        if bonus_guess == bonus_word:
            print("✅ Great job on the bonus!\n")
            score += .5
        else:
            print(f"❌ Oops! That one was '{bonus_word}'.\n")

# 🏁 Final Score
print(f"⭐ Final Score: {score}/{num_questions} ⭐")
```
